# House Prices — An Overview of Encoding Techniques

This notebook studies the categorical **encoding techniques** demonstrated in the
reference notebook (*"An overview of encoding techniques"*) and applies them to a
real dataset: the Kaggle **House Prices – Advanced Regression Techniques** dataset.



## Importing required libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split, KFold
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error

pd.set_option('display.max_columns', 50)
```

## Loading the data

```python
train = pd.read_csv('data/train.csv')
test = pd.read_csv('data/test.csv')

print('train data set has got {} rows and {} columns'.format(train.shape[0], train.shape[1]))
print('test data set has got {} rows and {} columns'.format(test.shape[0], test.shape[1]))
train.head()
```

### Defining the features and the target
The target we're trying to predict is `SalePrice`.

```python
plt.hist(train['SalePrice'], bins=40)
plt.title('distribution of target variable')
plt.show()
```

It's skewed (most houses are cheaper, a few are very expensive), so I'll predict `log(SalePrice)` instead of the raw price - this is what's usually done for this dataset, it makes the errors more balanced.

```python
y = np.log1p(train['SalePrice'])
X = train.drop(['SalePrice', 'Id'], axis=1)
```

Data variables present in this dataset:
- **Numerical data**: things like `LotArea`, `GrLivArea` - actual measurements.
- **Nominal data**: categories with no real order, like `Neighborhood`, `RoofStyle`.
- **Ordinal-ish data**: categories that do have some order, like `ExterQual` (Poor < Fair < Average < Good < Excellent), even though it's stored as text.
- **Cyclic data**: `MoSold` (month sold, 1-12), where month 12 is actually close to month 1.

```python
cat_cols = X.select_dtypes(include='object').columns.tolist()
num_cols = X.select_dtypes(exclude='object').columns.tolist()

print('categorical columns:', len(cat_cols))
print('numeric columns:', len(num_cols))
```

Before encoding anything I need to deal with missing values, otherwise the encoders will throw errors. Numeric columns get filled with the median, categorical columns get filled with the word 'Missing'.

```python
X_clean = X.copy()

for col in num_cols:
    X_clean[col] = X_clean[col].fillna(X_clean[col].median())

for col in cat_cols:
    X_clean[col] = X_clean[col].fillna('Missing')
```

## Baseline (numeric columns only, no categorical data)
Just so I have something to compare the encoded versions against.

```python
X_train, X_valid, y_train, y_valid = train_test_split(X_clean[num_cols], y, test_size=0.2, random_state=42)

model = LinearRegression()
model.fit(X_train, y_train)
preds = model.predict(X_valid)

rmse_baseline = np.sqrt(mean_squared_error(y_valid, preds))
print('Baseline RMSE:', rmse_baseline)
```

## Method 1: Label encoding
In this method every category gets substituted with a number. For example `RoofStyle`: Gable -> 0, Hip -> 1, and so on.

To do this we:
- Initialize the LabelEncoder class
- Call fit() on the data
- Transform the data into label-encoded data

```python
from sklearn.preprocessing import LabelEncoder

X_label = X_clean.copy()
for col in cat_cols:
    le = LabelEncoder()
    X_label[col] = le.fit_transform(X_label[col])

print('train data set has got {} rows and {} columns'.format(X_label.shape[0], X_label.shape[1]))
X_label.head()
```

```python
X_train, X_valid, y_train, y_valid = train_test_split(X_label, y, test_size=0.2, random_state=42)

model = LinearRegression()
model.fit(X_train, y_train)
preds = model.predict(X_valid)

rmse_label = np.sqrt(mean_squared_error(y_valid, preds))
print('Label Encoding RMSE:', rmse_label)
```

## Method 2: One-hot encoding
Here each category becomes its own column of 0s and 1s instead of one number. This avoids the false-order problem from Label Encoding, but adds a lot more columns - `Neighborhood` alone has 25 categories, so it turns into 25 columns.

To implement this we use the `get_dummies()` function in pandas.

```python
X_onehot = pd.get_dummies(X_clean, columns=cat_cols, drop_first=True)

print('train data set has got {} rows and {} columns'.format(X_onehot.shape[0], X_onehot.shape[1]))
```

```python
X_train, X_valid, y_train, y_valid = train_test_split(X_onehot, y, test_size=0.2, random_state=42)

model = LinearRegression()
model.fit(X_train, y_train)
preds = model.predict(X_valid)

rmse_onehot = np.sqrt(mean_squared_error(y_valid, preds))
print('One-Hot Encoding RMSE:', rmse_onehot)
```

## Method 3: Feature Hashing (the hashing trick)
Feature hashing represents categories in a "one-hot style" sparse format but with much lower dimensions. A hashing function is applied to each category and the result decides which column gets a 1. So instead of 260 one-hot columns, we can squeeze everything into a fixed, much smaller number of columns.

The trade-off is that two different categories can occasionally land in the same column (a "collision"), which loses a small bit of information.

```python
from sklearn.feature_extraction import FeatureHasher

cat_records = X_clean[cat_cols].astype(str).to_dict(orient='records')

hasher = FeatureHasher(n_features=50, input_type='dict')
hashed = hasher.transform(cat_records)
hashed_df = pd.DataFrame(hashed.toarray(), columns=[f'hash_{i}' for i in range(50)])

X_hash = pd.concat([X_clean[num_cols].reset_index(drop=True), hashed_df], axis=1)

print('train data set has got {} rows and {} columns'.format(X_hash.shape[0], X_hash.shape[1]))
X_hash.head(3)
```

```python
X_train, X_valid, y_train, y_valid = train_test_split(X_hash, y, test_size=0.2, random_state=42)

model = LinearRegression()
model.fit(X_train, y_train)
preds = model.predict(X_valid)

rmse_hash = np.sqrt(mean_squared_error(y_valid, preds))
print('Feature Hashing RMSE:', rmse_hash)
```

## Method 4: Encoding categories with dataset statistics
Now we give each category a numeric representation using a small number of columns, but in a way that puts similar categories close to each other. The simplest way to do this is to replace every category with how many times it appears in the dataset - if two neighborhoods are both common, they'll end up with similar numbers.

```python
X_stat = X_clean.copy()

for col in cat_cols:
    counts = X_stat[col].value_counts()
    X_stat[col] = X_stat[col].map(counts)

print('train data set has got {} rows and {} columns'.format(X_stat.shape[0], X_stat.shape[1]))
X_stat.head(3)
```

```python
X_train, X_valid, y_train, y_valid = train_test_split(X_stat, y, test_size=0.2, random_state=42)

model = LinearRegression()
model.fit(X_train, y_train)
preds = model.predict(X_valid)

rmse_stat = np.sqrt(mean_squared_error(y_valid, preds))
print('Dataset Statistics Encoding RMSE:', rmse_stat)
```

## Encoding cyclic features
Some features are cyclic in nature, like `MoSold` (month sold, 1-12). Month 12 (December) is actually close to month 1 (January), but a plain number doesn't capture that. A common way to handle this is to transform the value into two dimensions using sine and cosine.

```python
X_cyclic = X_stat.copy()
X_cyclic['MoSold_sin'] = np.sin(2 * np.pi * X_clean['MoSold'] / 12)
X_cyclic['MoSold_cos'] = np.cos(2 * np.pi * X_clean['MoSold'] / 12)
X_cyclic = X_cyclic.drop('MoSold', axis=1)

X_cyclic[['MoSold_sin', 'MoSold_cos']].head(3)
```

```python
X_train, X_valid, y_train, y_valid = train_test_split(X_cyclic, y, test_size=0.2, random_state=42)

model = LinearRegression()
model.fit(X_train, y_train)
preds = model.predict(X_valid)

rmse_cyclic = np.sqrt(mean_squared_error(y_valid, preds))
print('Dataset Statistics + Cyclic RMSE:', rmse_cyclic)
```

## Method 5: Target encoding
In this method we replace each category with the average target value for that category (average `SalePrice`, in log form). So if houses in a certain neighborhood tend to sell high, that neighborhood gets a high number.

**Note:** doing it this way, on the whole dataset at once, technically leaks information - each row's own price is included in the average used for its own category. I did it anyway first to see what happens, then fixed it properly in Method 6.

```python
X_target = X_clean.copy()
X_target['target_temp'] = y.values

for col in cat_cols:
    avg_price = X_target.groupby(col)['target_temp'].mean()
    X_target[col] = X_target[col].map(avg_price)

X_target = X_target.drop('target_temp', axis=1)
X_target.head(3)
```

```python
X_train, X_valid, y_train, y_valid = train_test_split(X_target, y, test_size=0.2, random_state=42)

model = LinearRegression()
model.fit(X_train, y_train)
preds = model.predict(X_valid)

rmse_target = np.sqrt(mean_squared_error(y_valid, preds))
print('Target Encoding RMSE (leaky):', rmse_target)
```

### K-Fold target encoding
K-Fold target encoding can be used to reduce the overfitting/leakage from Method 5. Here we split the data into 5 folds. For each fold, the category averages are calculated using only the *other* 4 folds, then applied to the held-out fold - so no row ever sees its own price used in its own encoding.

```python
X_fold = X_clean.copy()
X_fold['target_temp'] = y.values

kf = KFold(n_splits=5, shuffle=True, random_state=42)

for col in cat_cols:
    new_col = np.zeros(len(X_fold))
    for train_ind, val_ind in kf.split(X_fold):
        avg_price = X_fold.iloc[train_ind].groupby(col)['target_temp'].mean()
        overall_avg = X_fold.iloc[train_ind]['target_temp'].mean()
        new_col[val_ind] = X_fold.iloc[val_ind][col].map(avg_price).fillna(overall_avg).values
    X_fold[col] = new_col

X_fold = X_fold.drop('target_temp', axis=1)
X_fold.head(3)
```

```python
X_train, X_valid, y_train, y_valid = train_test_split(X_fold, y, test_size=0.2, random_state=42)

model = LinearRegression()
model.fit(X_train, y_train)
preds = model.predict(X_valid)

rmse_kfold = np.sqrt(mean_squared_error(y_valid, preds))
print('K-Fold Target Encoding RMSE:', rmse_kfold)
```

# Summary
Here's the summary of model performance for each encoding technique.

```python
results = pd.DataFrame({
    'Encoding': ['Baseline (numeric only)', 'Label Encoding', 'One-Hot Encoding', 'Feature Hashing',
                 'Dataset Statistics Encoding', 'Dataset Statistics + Cyclic',
                 'Target Encoding (leaky)', 'K-Fold Target Encoding'],
    'RMSE': [rmse_baseline, rmse_label, rmse_onehot, rmse_hash, rmse_stat, rmse_cyclic, rmse_target, rmse_kfold]
})

results = results.sort_values('RMSE')
results
```

```python
plt.figure(figsize=(8,4))
plt.barh(results['Encoding'], results['RMSE'])
plt.xlabel('RMSE (lower is better)')
plt.title('Comparing encoding methods')
plt.gca().invert_yaxis()
plt.tight_layout()
plt.show()
```

**What I noticed:** plain Target Encoding got the lowest error, but that's a bit misleading since it's leaking information from the target. K-Fold Target Encoding is a bit higher but more trustworthy since it doesn't cheat. One-Hot actually did worse here even though it doesn't lose any information,probably because it added so many columns that the model had a harder time fitting well. Feature Hashing did reasonably considering it compressed everything into just 50 columns. Label Encoding was quick but not very accurate, since it's a linear model and the fake ordering it creates works against it.
